```python
import numpy as np
import matplotlib.pyplot as plt

def round_to_half(x):
    return np.round(2*x)/2

N = 100000000
sigma = 0.50

T_T, T_R, T_S = 3.5, 3.0, 3.5
T_T1, T_R1, T_S1 = 3.0, 3.5, 3.0

eps_T = np.random.normal(0, sigma, N)
eps_R = np.random.normal(0, sigma, N)
eps_S = np.random.normal(0, sigma, N)
eps_T1 = np.random.normal(0, sigma, N)
eps_R1 = np.random.normal(0, sigma, N)
eps_S1 = np.random.normal(0, sigma, N)

S_T = round_to_half(T_T + eps_T)
S_R = round_to_half(T_R + eps_R)
S_S = round_to_half(T_S + eps_S)
S_T1 = round_to_half(T_T1 + eps_T1)
S_R1 = round_to_half(T_R1 + eps_R1)
S_S1 = round_to_half(T_S1 + eps_S1)

S_overall = 0.50*S_T + 0.30*S_R + 0.20*S_S
S_overall1 = 0.50*S_T1 + 0.30*S_R1 + 0.20*S_S1

```

```python
# load the histogram data for both faculty members
hist1, bins1 = np.histogram(S_overall, bins=41, range=(2, 4))
hist2, bins2 = np.histogram(S_overall1, bins=41, range=(2, 4))
# plot the histogram data for both faculty members
plt.plot(bins1[:-1], hist1, drawstyle='steps-post', label='Faculty Member A')
plt.plot(bins2[:-1], hist2, drawstyle='steps-post', label='Faculty Member B')

plt.xlabel("Overall Score")
plt.ylabel("Frequency")
plt.title("Monte Carlo Distribution of Overall Score")
# Add vertical lines for the means
plt.axvline(np.mean(S_overall), color='blue', linestyle='dashed', linewidth=1, label='Mean A: {:.2f}'.format(np.mean(S_overall)))
plt.axvline(np.mean(S_overall1), color='orange', linestyle='dashed', linewidth=1, label='Mean B: {:.2f}'.format(np.mean(S_overall1)))
plt.legend()
# Set ymax to 15000000
#plt.ylim(0, 15000000)
# Set x limits to 0 to 4
plt.xlim(2, 4)
plt.show()

print("Mean1:", np.mean(S_overall))
print("Std dev1:", np.std(S_overall))
print("Mean2:", np.mean(S_overall1))
print("Std dev2:", np.std(S_overall1))
```
